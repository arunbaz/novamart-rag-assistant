```python
returns_policy = """
# NovaMart Australia — Returns and Refunds Policy

## Change-of-mind returns

Customers may return an unused product within 30 calendar days of delivery.

The product must:

- Be unused and in its original condition
- Include the original packaging, accessories and manuals
- Include proof of purchase

Change-of-mind returns are not accepted after 30 days.

Customers are responsible for return shipping costs for change-of-mind returns.

## Damaged or incorrect products

Customers must contact NovaMart Support within 7 calendar days of delivery if a product arrives damaged, defective or incorrect.

NovaMart will provide a prepaid return label. The customer may choose:

- A replacement
- A repair, where appropriate
- A full refund to the original payment method

Customers should provide their order number and clear photos of the product and packaging.

## Refund processing

NovaMart will inspect returned products after receiving them.

Approved refunds are issued to the original payment method within 5–7 business days. The customer’s bank may require additional time to display the refund.

## Non-returnable products

The following products cannot be returned for change of mind:

- Personalised products
- Opened hygiene products
- Digital gift cards
- Clearance products marked as final sale

This policy does not limit customers’ rights under Australian Consumer Law.
"""

file_path = project_root / "data" / "knowledge_base" / "returns_refunds.md"
file_path.write_text(returns_policy.strip(), encoding="utf-8")

print(f"Created: {file_path}")
```

```python
shipping_policy = """
# NovaMart Australia — Shipping and Delivery Policy

## Delivery locations

NovaMart delivers to residential and business addresses across Australia. International delivery is not currently available.

## Standard delivery

Standard delivery costs $7.95.

Estimated delivery times are:

- Metropolitan areas: 3–5 business days
- Regional areas: 5–8 business days
- Remote areas: 7–12 business days

Standard delivery is free for orders valued at $100 or more.

## Express delivery

Express delivery costs $14.95.

Estimated delivery times are:

- Metropolitan areas: 1–2 business days
- Regional and remote areas: 2–5 business days

Express delivery may not be available for oversized products or some remote locations.

## Order processing

Orders placed before 2:00 PM AEST on a business day are usually processed on the same day.

Orders placed after 2:00 PM, on weekends or on public holidays are processed on the next business day.

## Tracking orders

Customers receive a shipping confirmation email containing a tracking link after the order leaves the warehouse.

Tracking information may take up to 24 hours to become active.

## Delayed or missing orders

Customers should contact NovaMart Support if:

- Tracking has not updated for 3 business days
- A standard order has not arrived after its estimated delivery period
- Tracking shows delivered but the parcel cannot be found

Customers should provide their order number when contacting support.
"""

file_path = project_root / "data" / "knowledge_base" / "shipping_delivery.md"
file_path.write_text(shipping_policy.strip(), encoding="utf-8")

print(f"Created: {file_path}")
```

```python
knowledge_base_path = project_root / "data" / "knowledge_base"
document_files = sorted(knowledge_base_path.glob("*.md"))

print(f"Total documents: {len(document_files)}\n")

for document in document_files:
    print(f"- {document.name}")
```

```python
documents = []

for file_path in document_files:
    document = {
        "content": file_path.read_text(encoding="utf-8"),
        "source": file_path.name
    }
    documents.append(document)

print(f"Loaded {len(documents)} documents.")
print(f"First source: {documents[0]['source']}")
print("\nPreview:")
print(documents[0]["content"][:500])
```

```python
chunks = []

for document in documents:
    sections = document["content"].split("\n## ")

    for section_number, section in enumerate(sections):
        if section_number > 0:
            section = "## " + section

        chunk = {
            "content": section.strip(),
            "source": document["source"],
            "chunk_id": f"{document['source']}_{section_number}"
        }

        chunks.append(chunk)

print(f"Original documents: {len(documents)}")
print(f"Created chunks: {len(chunks)}")

print("\nExample chunk:")
print(f"Chunk ID: {chunks[1]['chunk_id']}")
print(f"Source: {chunks[1]['source']}")
print(chunks[1]["content"])
```

```python
chunks = []

for document in documents:
    sections = document["content"].split("\n## ")
    document_title = sections[0].strip()

    for section_number, section in enumerate(sections[1:], start=1):
        chunk_text = f"{document_title}\n\n## {section.strip()}"

        chunk = {
            "content": chunk_text,
            "source": document["source"],
            "chunk_id": f"{document['source']}_{section_number}"
        }

        chunks.append(chunk)

print(f"Original documents: {len(documents)}")
print(f"Useful chunks created: {len(chunks)}")

print("\nExample improved chunk:")
print(f"Chunk ID: {chunks[0]['chunk_id']}")
print(f"Source: {chunks[0]['source']}")
print(chunks[0]["content"])
```

```python
import json

processed_data_path = project_root / "data" / "processed"
processed_data_path.mkdir(parents=True, exist_ok=True)

chunks_file = processed_data_path / "knowledge_base_chunks.json"

with chunks_file.open("w", encoding="utf-8") as file:
    json.dump(chunks, file, ensure_ascii=False, indent=2)

print(f"Saved {len(chunks)} chunks to:")
print(chunks_file)
```
